```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.getOrCreate()

# Read CSV using pandas from workspace
diabetic_data = pd.read_csv('/Workspace/Users/rishi.khandelwal@fractal.ai/MLOps/diabetic_data.csv')
IDS_mapping = pd.read_csv('/Workspace/Users/rishi.khandelwal@fractal.ai/MLOps/IDS_mapping.csv')

# Convert to Spark DataFrame
diabetic_data_df = spark.createDataFrame(diabetic_data)
IDS_mapping_df = spark.createDataFrame(IDS_mapping)

diabetic_data_df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable('workspace.default.diabetic_data')

IDS_mapping_df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable('workspace.default.ids_mapping')
```

```python
df = spark.table('diabetic_data')
display(df)
```
